## What Is Algebra? 

By now, you should already be familiar with arithmetic. In arithmetic,
we only deal with numbers and mathematical operations on them like
addition, subtraction, multiplication and division. The knowledge of
arithmetic is very useful for solving simple problems but it will not be
sufficient for more general and difficult problems. For example, if we
know that a number and its half added together become 15, what is that
number? You may well know the answer to this question, but without some
basic knowledge of algebra it will probably take you some time to find
it. So what is algebra? Basically, algebra is a branch of mathematics
that has been developed to manipulate symbols. In algebra we use letters
(English or Greek alphabets for example) to represent numbers or
quantities of interest. In this chapter, first we will look at how to
create simple algebraic expressions from given contexts and then look at
some techniques that we can use to simplify them.

Symbols used in algebra essentially represent quantities (with or
without units) that could be known already or unknown yet. They can be
constants, or variables, whose values can change from case to case.

For example, we can use $g=9.8~ \mathrm{m/s}^2$ to represent the
gravitational acceleration on the Earth, we can use the letter $n$ to
represent the number of students in a class.

## Creating Algebraic Expressions

Why do we use letters to stand for quantities?

Let’s look at some examples

```{prf:example}
Errol buys
$\boldsymbol{x}$ apples and $\boldsymbol{y}$ oranges. Apples are 39p
each and oranges are 56p each. How much does he spend? 
$$\left .% \{
      \begin{tabular}{ll}
      \text{Total spent} =& ~ \text{the number of apples that Errol buys times 39 pence} \\
       &  +  \text{the number of oranges that Errol buys times 56 pence} 
      \end{tabular}
    \right \}
    \label{eq:1.1}$$ 
If we use the letters $\boldsymbol{x}$ and
$\boldsymbol{y}$ to represent, respectively, the number of apples and
the number of oranges that Errol buys, we can write the total spent as
$$39x+56y \quad \text{(units in pence)}
    \label{eq:1.2}$$
```


Suppose that Errol buys $\boldsymbol{z}$ pineapples as well and they are
£1.6 each, the total cost is now
$$39x+56y+160z  \quad \text{(units in pence)}
    \label{eq:1.3}$$

The £1.6 has to be expressed as 160p so that we are adding together
pence throughout the whole expression. In general, we can only add
things of the same type.

When internet calls at off-peak time are charged at 3p per minute then
the cost of a call can be calculated from the formula
$$\text{Cost (in p)} = 3 \times \text{time on call (in minutes)}$$ This
ensures that if we double the time on call we will double the cost. For
instance the cost of a 20 minute call is 60p. If we double the time to
40 minutes the call doubles to 120p.

In mathematics we could reduce this to $$\label{eq:1.4}
    C=3\times t$$ Remember that the symbols represent the cost and the
time and the corresponding units are pence and minutes.

These are examples of the simplest form of **algebraic notation**. The
abbreviation using symbols saves time and enables us to concentrate on
the important issues associated with the formula. In fact in algebra
when two values are placed side by side we assume they are to be
multiplied so we would normally write the formula in
<a href="#eq:1.4" data-reference-type="eqref"
data-reference="eq:1.4">[eq:1.4]</a> above as $$C=3t$$

The quantities $\boldsymbol{x}$, $\boldsymbol{y}$ and $\boldsymbol{z}$
in <a href="#eq:1.2" data-reference-type="eqref"
data-reference="eq:1.2">[eq:1.2]</a> and
<a href="#eq:1.3" data-reference-type="eqref"
data-reference="eq:1.3">[eq:1.3]</a> for cost are **variables**; they
stand for any one of a range of possible values. A variable may be given
a specific value when we want to work out a particular sum.

Similarly, in <a href="#eq:1.4" data-reference-type="eqref"
data-reference="eq:1.4">[eq:1.4]</a>, the values for $t$ and $C$ will
change from call to call and so they are also called variables. We tend
to think we can choose the time to be any value within reason. The cost
then depends on the value we have chosen. Because of this we call $t$
the **independent** variable and $C$ the **dependent** variable.

<a href="#eq:1.2" data-reference-type="eqref"
data-reference="eq:1.2">[eq:1.2]</a> defines a function for calculating
the cost in terms of the variables $\boldsymbol{x}$ and
$\boldsymbol{y}$, which works whatever values are given to
$\boldsymbol{x}$ and $\boldsymbol{y}$.

We use variables and functions to describe general properties. This has
huge advantages over specifying properties in a multitude of individual
specific cases. From the general cases, we can establish patterns which
can be used to solve other similar problems.

Because variables represent numbers, we manipulate expressions involving
variables using the rules of arithmetic.


::::{exercise}
:label: exercise:1.1
1.  Write algebraic expression (equation) for each the following:

    1.  The number that is 4 more than $\boldsymbol{x}$ is 7

    2.  The number that is twice $\boldsymbol{x}$ is 5.

    3.  The number that is half of $\boldsymbol{y}$ is 8.

    4.  The number that is 6 less than $\boldsymbol{y}$ is 4.

    5.  The number that is a quarter of $\boldsymbol{x}$ is 2.

2.  Determine the appropriate formulae in each of the following cases
    and write it down in abbreviated form using symbols. State clearly
    what the symbols represent and the corresponding units.

    1.  The cost of a rope is equal to its length in metres times 90p
        per metre.

    2.  The distance travelled on a journey in miles is equal to the
        average speed multiplied by the time taken.

    3.  The concentration of a substance in solution is equal to the
        mass of the substance divided by the volume of the solution.

    4.  The area of a triangle is equal to half the base times the
        perpendicular height.

    5.  The cost in pence of an electricity bill is the standing charge
        plus 0.9 times the number of units used.

3.  Evaluate the following algebraic expressions when $x=10$ and $y=5$. 

    1.  $x+y$

    2.  $x-y$

    3.  $xy$

    4.  $\dfrac{x}{y}$

4.  Susan buys 7 apples at $\boldsymbol{x}$ pence each and 4 bananas at
    $\boldsymbol{y}$ pence each. She spends £1. Write an equation for
    this.

5.  Jarred pays £$\boldsymbol{x}$ for a bus ticket. The following week
    the price has risen by £2. In total he spent £$\boldsymbol{T}$ on
    the two bus tickets. Write an equation expressing this information.

6.  Fred runs $\boldsymbol{x}$ miles and John runs 3 times as far. John
    runs $\boldsymbol{y}$ miles. Write an equation that relates
    $\boldsymbol{x}$ and $\boldsymbol{y}$.
    
::::

## Substitution and BODMAS

In the previous section, we have created a number of algebraic
expressions. In Example <a href="#example:1A" data-reference-type="ref"
data-reference="example:1A">[example:1A]</a>, the cost for Errol to buy
$\boldsymbol{x}$ apples and $\boldsymbol{y}$ oranges is represented by
the algebraic expression $$39x+56y.$$ If at one time he buys 5 apples
and 4 oranges, how much will he pay? To work out this we simply replace
$\boldsymbol{x}$ with number 5 and $\boldsymbol{y}$ with 4 and then the
total cost will be
$$39 \times  5+56 \times 4=195+216=419~ \text{(pence) or} ~4.19~ \text{(pounds)} .$$

This process of replacing letters (symbols) in an algebraic expression
with actual numbers and then working out the final numerical value is
called **evaluation by substitution**.

Evaluate $2x+4y+10$ when $x=1.4$, $y=2.2$.

Substituting $x=1.4$ and $y=2.2$ into $2x+4y+10$, we have
$$\begin{aligned}
    & 2x+4y+10 \\
    =~ & 2\times 1.4+4\times 2.2+10  \\
    =~ & 2.8 + 8.8 + 10\\
    =~ & 21.6
\end{aligned}$$

In algebra we use letters or symbols to represent numbers **so the
operations on them must follow the same rules that numbers obey.**  The
BODMAS rule tells us the order in which the mathematical operations must
performed.

| **BIDMAS**         |                 | **BODMAS**                  |
|:-------------------|:---------------:|:----------------------------|
| **B**rackets       | First priority  | **B**rackets                |
| **I**ndex          | Second priority | Exp**o**nents and ro**o**ts |
| **D**ivision       | Third priority  | **D**ivision                |
| **M**ultiplication | Third priority  | **M**ultiplication          |
| **A**ddition       | Fourth priority | **A**ddition                |
| **S**ubtraction    | Fourth priority | **S**ubtraction             |

If an expression contains two operations with the same priority, we
evaluate by working from left to right.

1.  To evaluate an expression such as $$11-3+14,$$ we would follow the
    **BODMAS** rule giving 22 as the answer.

    Since addition and subtraction are of equal priority we scan the
    expression from left to right calculating as we go.

    Similarly to find the value of  $a+b-c$  we first add $a$ to $b$ and
    then subtract $c$.

2.  When dealing with a numerical expression such as $$3+5 \times 2$$ we
    perform the multiplication before the addition (to get 13).

    So when evaluating the algebraic expression $$a+bc,$$ we would find
    the value of $b \times c$ first and then add the value of $a$.

3.  To override the normal rules of precedence we use **brackets**.  
    Thus to calculate $$6 \times (2+8)$$ we perform the addition before
    the multiplication. This gives an answer of 60.  
    Likewise to find the value of $$a(b+c)$$ we first add $b$ to $c$ and
    then multiply the result by $a$.

Since algebraic symbols only follow the rules of numbers if a rule does
not work for numbers, it does not work for algebra either.

To add fractions we do not simply add the denominators and the
numerators. This would give us the wrong answer. For instance,

$\dfrac 1 2+\dfrac 1 3$ does not equal $\dfrac{1+1}{2+3}=\dfrac 2 5$

so in algebra $\dfrac a c+\dfrac b d$ is not equal to $\dfrac{a+b}{c+d}$

<span id="exercise:1.2" label="exercise:1.2"></span> Evaluate the
following expressions when $w=3$, $x=5$, $y=12$ and $z=2$ without using
a calculator

2

1.  $10x$

2.  $2y-11$

3.  $w+2z$

4.  $(w+2)z$

5.  $17-3w$

6.  $(x+1)(y-7)$

7.  $3x-y+z$

8.  $5x+3z-3x$

9.  $\dfrac y w$

10. $\dfrac{x+3} z$

11. $\dfrac 1 x+\dfrac 1 w$

## Powers in Algebra

In Arithmetic, when a number is multiplied to itself several times, we
can use a short-hand form called power or index to represent it. For
example, $$2\times 2\times 2\times 2$$ can be written as $2^4$ (2 raised
to the power of 4). Similarly, in algebra we write $a^2$ for the product
$a\times a$,   $a^3$ for $a\times a\times a$, and so on. The calculation
of a power takes precedence over multiplication and division.

To find the value of $5a^2$ we first calculate a<sup>2</sup> and then
multiply by 5 so, if $a=3$, $a^2=9$ and $5a^2=45$.

If we had wanted to perform the multiplication first, we would have
written $(5a)^2$. Since $a=3$, $5a=15$ and

$(5a)^2=15\times 15=225$.

Take care here. Failing to make the calculations in the right order is a
very common error.

<span id="exercise:1.3" label="exercise:1.3"></span>

1.  Find the value of the following expressions when $x=3$ and $y=2$.

    2

    1.  $3x^2$

    2.  $(3x)^2$

    3.  $3y^2$

    4.  $(x+y)^2$

    5.  $xy^2$

    6.  $(xy)^2$

2.  Write down the algebraic expression corresponding to the following
    instructions expressed in words.

    1.  Multiply $\boldsymbol{u}$ by 6 and then cube the result.

    2.  Halve the square of $\boldsymbol{v}$.

    3.  Add $\boldsymbol{p}$ to twice of $\boldsymbol{q}$ and square the
        result.

    4.  Subtract $\boldsymbol{a}$ cubed from $\boldsymbol{b}$ squared.

## Roots

When a number multiplies itself, we can use power to represent the
operation, for example $$3\times 3 = 3^2 = 9$$ Here we know $9$ is a
square number. And we call $3$ a square root of $9$, and write this as
$$\sqrt[2]{9}=\sqrt{9}=3$$ At the same time
$$(-3)\times (-3) = (-3)^2=9,$$ so $-3$ is also a square root of $9$.

A positive number has two opposite square roots. For example $5$ and
$-5$ are the square roots of $25$. A real number has only one cube root.

In general, if we have
$$\underbrace{a \times a \times a \cdots a \times a }_{n} = a^n = b$$
then the $n^\text{th}$ root of $b$ is $$\left\{
\begin{aligned}
     ~  \sqrt[n]{b},  \quad & n~ \text{is odd}, \quad & b \in \mathbb{R} \\
     ~  \pm\sqrt[n]{b}, \quad & n~ \text{is even}, \quad & b \geq 0 \\
\end{aligned}
    \right.$$ We can write the $n^\text{th}$ root of $b$ as
$$\sqrt[n]{b}=b^{\frac{1}{n}},$$ so

-   $\sqrt{9}=9^{\frac{1}{2}}$

-   $\sqrt[3]{8}=8^{\frac{1}{3}}$

-   $\sqrt[10]{x}=x^{\frac{1}{10}}$

## Gathering Like Terms and Expanding Brackets

Algebraic expressions can sometimes be shortened by combining terms. One
instance of this can arise when the same term appears twice in a sum.
Thus we can simplify

$a+a+9$  to  $2a+9$

We can still simplify the expression if one or both of these terms are
multiplied by a constant or if the terms are subtracted.

1.  $2a=a+a$  and  $3a=a+a+a$ ${\Rightarrow}$  $2a+3a=5a$

2.  $6b=b+b+b+b+b+b$  and  $4b=b+b+b+b$  ${\Rightarrow}$ $6b-4b=2b$ 

We call this process **“gathering like terms”**. Algebraic expressions
may contain two or more sets of like terms and constants also.

In the expression   $8x+5y+2+3x-y-3$

$8x$ and  $3x$ are like terms  and 5y and y are like terms. Gathering
like terms together we get $$8x+3x+5y-y+2-3$$ which simplifies to
 $11x+4y-1$.

Be careful not to combine unlike terms in this way. We cannot simplify
$11x+4y-1$ any further.

<span id="exercise:1.4" label="exercise:1.4"></span> Simplify the
following expressions by gathering like terms where possible.

1.  $6p+3$

2.  $10q-4q$

3.  $7x+4y-2x+3y$

4.  $4z^2+z$

5.  $a+3b+2c+4a-5b$

6.  $4s^2+8r^2+1+s^2-3r^2+3$

7.  $12u+8v+5w+6u+7v-5w+1$

8.  $xy+xz+yz$

When creating algebraic expressions, frequently we use brackets to group
some quantities together. To simplify certain algebraic expressions,
such as, $1.2(2x+4y)$, sometimes we need to remove all the brackets in
them first. To do this, you need to use the number or letter in front of
the brackets to *multiply every term inside them*. This process is
called **expanding the brackets**. For example, after expanding the
brackets, the above expression will become $2.4x+4.8y$. Notice that the
$1.2$ multiplies every term in the brackets not just the first one. This
remains true when there are more than two terms in the brackets and when
the terms are negative.

1.  $2(p+q+r)=2p+2q+2r$ (Not $2p+q+r$)

2.  $5(2x^2-y^2)=10x^2-5y^2$

When there is a negative term outside the brackets, it is important to
remember that the product of two negative terms is positive. This
includes the case where there is only a subtraction sign preceding the
brackets, as in Example <a href="#example:1K" data-reference-type="ref"
data-reference="example:1K">[example:1K]</a> (d) below. In practice,
this implies there is $-1$ before the brackets. Getting the signs wrong,
when expanding brackets, is another very common error.

Once we have expanded two or more sets of brackets we may be able to
gather like terms.

<span id="example:1K" label="example:1K"></span>

1.  $\begin{aligned}[t] 9(2x+y)-7(x-y) & = 18x+9y-7x+7y \\
                & = 18x-7x+9y+7y \\
                & = 11x+16y            
            \end{aligned}$

2.  $\begin{aligned}[t]
                13x-(x-7) & = 13x-x+7 \\
                          & = 12x+7
            \end{aligned}$

<span id="exercise:1.5" label="exercise:1.5"></span> Expand the brackets
in the following expressions, simplifying your answer where possible.

1.  $4(x+y)$

2.  $10(3a-5b)$

3.  $6(u+v-20)$

4.  $2(4a+3b)+3(a+3b)$

5.  $5(2s+t)-2(2s-t)$

6.  $8(p+2q-2r)-7(p-2q-3r)$

7.  $4x+5y-(2x-3y+5)$

8.  $6(x^2+y^2+1)-(x^2+y^2)-2(x^2-2y^2-1)$

9.  $x^2(x^2-2x)$

10. $x^2(x^2-5x+2)$

11. $x(2x+1)$

12. $x(-x+1)-2x(3x+1)$

13. $2x^2(x-3)+5x(x^2+2x-1)$

## Simple Factorisation

The reverse process of ‘expanding the brackets’ is called
**factorising**. An example will suffice.

If we expand $10(2x+3y)$ we get   $20x+30y$.  
If, however, we start with the expression $20x+30y$ and rewrite it as
the product of two factors

 i.e.  as $10(2x+3y)$,

we say the expression has been factorised.

We can only factorise terms when they have a **common factor**, for
instance, we could not factorise $15x+14y$ because none of the factors
of the first term, 3, 5 and $x$ are factors of the second term.

Spotting the common factor is the essential first step in the
factorisation process. Once the process is completed we can always check
our answer by expanding the brackets to see if we obtain the original
expression.

1.  All the terms in the expression  $3x+6y-18z$  have a factor 3.  
    To find the second factor we divide each term by 3 giving us
    $x+2y-6z$.  
    Hence we believe that $$3x+6y-18z=3(x+2y-6z).$$ If we now expand the
    brackets we get $3x+6y-18z$, so our factorisation is indeed correct.

2.  The term $2x$ is common to all the terms in the expression
     $8xy+6x^2$.  
    Dividing each of the terms by $2x$ we get $4y+3x$. Hence we think
    that $$8xy+6x^2=2x(4y+3x).$$

<span id="exercise:1.6" label="exercise:1.6"></span> Factorise the
following expressions.

1.  $7u+21v$

2.  $16x+24y$

3.  $z^2+3z$

4.  $8p-20q+12r$

5.  $2a^2-11ab$

6.  $49xyz+8yz$

Check your answers by expanding them.

## Further Expansions

We normally write the product of two or more sets of terms in brackets
without the multiplication sign. Thus the expression $(x+2)\times (x+3)$
is written $(x+2)(x+3)$. When expanding such an expression involving two
pairs of brackets we must multiply every term in the first bracket by
every term in the second.

### General Method

-   **Two brackets**:  
    The general rule for expanding two brackets is that *every term in
    the first bracket should multiply every term in the second bracket*.
    Of course, we can also treat one bracket as a single symbol, then
    deal with the expression as one bracket multiplying one symbol. For
    example $$\begin{aligned}
        (a+b)(c+d)&=a(c+d)+b(c+d) \\
                  &=ac+ad+bc+bd
        
    \end{aligned}$$ here we treat $(c+d)$ as a single symbol first, and
    let it multiply every term in $(a+b)$.

-   **Multiple brackets**:  
    To expand multiple brackets, we can iteratively expand two brackets.
    For example $$\begin{aligned}
        (a+b)(c+d)(e+f+g)
              ~=~&(ac+ad+bc+bd)(e+f+g)  \\
              ~=~&ace+ade+bce+bde \\
              ~&~+acf+adf+bcf+bdf \\
              ~&~+ acg + adg + bcg +bdg
        
    \end{aligned}$$ here we firstly expand the first two brackets
    $(a+b)(c+d)$, then expand the remaining two brackets
    $(ac+ad+bc+bd)(e+f+g)$.

### Box Method

We can justify the process and determine the answer using a diagram of a
rectangle.

We can see that  

$$\begin{aligned}
(x+2)(x+3)& = (\text{length}) \times (\text{height of rectangle}) \\
         & = \text{the area of the rectangle}  \\
         & =x^2+2x+3x+6,    
\end{aligned}$$

c\|c\|c\| & &  
$x$ & $x^2$ & $2x$  
$3$ & $3x$ & $6$  

and gathering like terms gives $$(x+2)(x+3)=x^2+2x+3x+6.$$

Again when multiplying the terms we need to remember that the product of
two negative terms is a positive one.

### ‘Smiley Face’ Method

$$\begin{aligned}
    &=x^2+6+2x+3x \\
    &=x^2+5x+6    
\end{aligned}$$

### ‘Moon’ Method

$$\begin{aligned}
    &=x^2+6+2x+3x \\
    &=x^2+5x+6    
\end{aligned}$$

$$\begin{aligned}
(x-3)^2 =&(x-3)(x-3)  \\
        =&x^2-3x-3x+9 \\
        =&x^2-6x+9      
\end{aligned}$$ Notice that $(x-3)^2$ is not equal to $x^2-9$.

<span id="exercise:1.7" label="exercise:1.7"></span> Simplify the
following expressions by expanding brackets

1.  $(x+5)(x+2)$

2.  $(x-2)(x+1)$

3.  $(x-10)(x-1)$

4.  $(y-4)(y+4)$

5.  $(z+5)^2$

6.  $(s-1)^2$

## Further Factorisation

### Factorising $x^2+Bx+C$

We described earlier how it was possible to expand an algebraic
expression and to conduct the reverse process of factorisation. For
instance we could expand the expression $u(3+v)$ to obtain $3u+uv$ and
we could factorise $3u+uv$ back to $u(3+v)$. On the last page we
considered the expansion of products of brackets. We showed that
$$(x+2)(x+3)=x^2+5x+6$$ and now, as you might expect, we discuss how we
might factorise an expression such as $$x^2+5x+6$$ back to the original
pair of brackets. This is not as easy a problem as one might expect.
Sometimes the factorisation is not possible, at least using ordinary
numbers. Even when factorisation is possible it is not always clear what
the original brackets were. However, there are clues. Consider the
following expansion $$(x+a)(x+b)=x^2+ax+bx+ab$$ We assume we are given
the expression on the right and wish to find the constants $a$ and $b$.

The expression can be written $x^2+x(a+b)+ab$ . The clues to the values
of $a$ and $b$ are therefore

1.  their sum is the term multiplying $x$, and

2.  their product is the constant term.

Factorise $x^2+6x+8$

By clue (ii) the product of $a$ and $b$ is $8$ so assuming they are
integers the possible pairs of brackets are

1.  $(x+1)(x+8)$

2.  $(x+2)(x+4)$

3.  $(x-1)(x-8)$

4.  $(x-2)(x-4)$

but by clue (i) the sum of $a$ and $b$ is $6$ . This is only true for
the second pair of brackets. Hence $$x^2+6x+8=(x+2)(x+4)$$

<span id="exercise:1.8" label="exercise:1.8"></span> Factorise each of
the following expressions.

1.  $x^2+4x+3$

2.  $x^2+7x+10$

3.  $x^2+8x+12$

4.  $x^2-7x+6$

5.  $x^2-x-6$

Check your answers by expanding them.

### Factorising $Ax^2+Bx+C$

Supposing a general quadratics $Ax^2+Bx+C$ can be factorised as
$$Ax^2+Bx+C=(ax+b)(cx+d),$$ we can expand the two brackets to find the
relations between the coefficients $$\begin{aligned}
    (ax+b)(cx+d)&=acx^2+adx+bcx+bd \\
                &=acx^2+(ad+bc)x+bd \\
                &=Ax^2 + B x + C.
\end{aligned}$$ We get the following clues:

1.  $A=ac$

2.  $C=bd$

3.  $B=ad+bc$

4.  $(ad)\cdot(bc)=A\cdot C = (ac)\cdot(bd)$

Using clues (iii) and (iv), we can work out the factorisation through
the following steps:

1.  Multiply $A$ and $C$, denote the result as $D=A\cdot C$

2.  Factorise $D$ as $D=B_1 \cdot B_2$, and make sure $B_1 + B_2 = B$

3.  Split $Bx$ as $Bx=B_1 x + B_2 x$

4.  Factorise $(Ax^2+B_1 x)+(B_2 x+C)$

Factorise $2x^2+5x+3$

**Solution**:

1.  $2\times3= 6$

2.  $6=1\times 6$, $6=2\times 3$, $2+3=5$

3.  $5x=2x+3x$

4.  $\begin{aligned}[t]
            2x^2+5x+3&=2x^2+2x+3x+3\\
                     &=(2x^2+2x)+(3x+3)\\
                     &=2x(x+1)+3(x+1)\\
                     &=(2x+3)(x+1)
        \end{aligned}$

Factorise $2x^2+3x-2$

**Solution**:

1.  $2\times(-2)= -4$

2.  $-4=1\times (-4)$, $-4=-1\times 4$, $-4=-2\times 2$, $-1+4=3$

3.  $3x=-x+4x$

4.  $\begin{aligned}[t]
            2x^2+3x-2&=2x^2-x+4x-2\\
                     &=(2x^2-x)+(4x-2)\\
                     &=x(2x-1)+2(2x-1)\\
                     &=(x+2)(2x-1)
        \end{aligned}$

<span id="exercise:Sec1.9.2" label="exercise:Sec1.9.2"></span> Factorise
each of the following expressions.

1.  $8x^2+18x+9$

2.  $6x^2+17x+12$

3.  $12x^2-8x-7$

4.  $15x^2+x-2$

5.  $12x^2-17x+6$

Check your answers by expanding them.

## Laws of Indices

The laws of indices listed below can be used for simplifying many
algebraic expressions involving powers. For examples, the first law says
that to multiply two numbers with the same base we can simply add the
indices and the second law states that to divide two numbers with same
base we can subtract the indices.

On the other hand, the sixth law gives the definition for negative
powers. According to this law, for example, $2^{-3}$ means
$\dfrac{1}{2^3}$ , that is $\dfrac{1}{8}$.

\|c\|p4cm\|l\|   & Laws of indices &  
.& $b^n \times b^m = b^{n+m}$ & $x^2 \times x^7= x^{2+7}=x^9$  
.& $b^n \div b^m = b^{n-m}$ & $x^{11} \div x^6= x^{11-6}=x^5$  
.& $(b^n)^m = b^{nm}$ & $(x^3)^4 = x^{3\times 4}= x^{12}$  
.& $b^1 =b$ & i.e. power 1 can be omitted.  
.& $b^0 =1$ & this gives the definition for zero power.  
.& $b^{-n}=\dfrac{1}{b^n}$ & $x^{-8} =\dfrac{1}{x^8}$  

<span id="exercise:1.9" label="exercise:1.9"></span>

1.  Use the laws of indices to simplify the following expressions:

    2

    1.  $a^{3} \times a^2$

    2.  $a^{3} \times a^{-1}$

    3.  $a^{31} \times a^{29}$

    4.  $a^{6} \div a^{4}$

    5.  $a^{9} \div a^{2}$

    6.  $(a^3)^4$

    7.  $a^3 \div a^{-2}$

    8.  $a^3 \times a^{-1} \times a^{5}$

    9.  $a^3 \times a^2 \div  a^7$

    10. $(a^{-2})^{-4}$

2.  Simplify the following expressions as far as possible:

    2

    1.  $2^2 \times 2^3$

    2.  $3^2 \times 3^3 \div 3^4$

    3.  $\dfrac{x^2 \times x^4}{x \times x^3}$

    4.  $(2^3)^4$

    5.  $\dfrac{x^3 \times (x^2)^5}{x^7}$

    6.  $\dfrac{(u^4)^{\frac{1}{2}}}{(u^{\frac{1}{2}})^6 \times u^{-2}}$

## Polynomial and Polynomial Division

A polynomial is an algebraic expression involving a sum of powers in one
or more variables multiplied by coefficients. A polynomial in one
variable (i.e., a univariate polynomial) with constant coefficients is
given by $$\label{eq:polynomial}
    a_n x^n + a_{n-1} x^{n-1} + + a_{n-2} x^{n-2} \cdots + a_2 x^2 + a_1 x + a_0.$$

-   In a polynomial, the indices should be **whole** numbers
    (non-negative integers) only.

-   The highest power in a univariate polynomial is called its
    **degree**, or sometimes its order.

-   $a_n$, $a_{n-1}$, $a_{n-2}$, …, $a_2$, $a_1$ and $a_0$ are constants
    and they are called the **coefficients** of the polynomial.

| Polynomial                                      | Degree | Name      |
|:------------------------------------------------|:-------|:----------|
| $a_0$                                           | 0      | constant  |
| $a_1 x + a_0$, $a_1 \neq 0$                     | 1      | linear    |
| $a_2 x^2 + a_1 x + a_0$, $a_2 \neq 0$           | 2      | quadratic |
| $a_3 x^3 + a_2 x^2 + a_1 x + a_0$, $a_3 \neq 0$ | 3      | cubic     |

| Algebraic expression    | Polynomial? |                                   |
|:------------------------|:-----------:|:----------------------------------|
| $4x^3+3x^2-4x+6$        |             |                                   |
| $x^7+3x^6+x^4+3x^2-x+2$ |             |                                   |
| $x^3+3x^2-4\sqrt{x}+6$  |             | $-4\sqrt{x}=-4x^\frac{1}{2}$      |
| $3x^2-9x+6$             |             |                                   |
| $3x^2+\dfrac{6}{x}-4$   |             | $\dfrac{6}{x}=6x^{-1}$            |
| $x^{-3}+3x^2+4$         |             |                                   |
| $x+\sin x$              |             | $\sin x$  is not a power function |

Similar to the division of numbers, we can also divide one polynomial by
another polynomial of the same or lower degree.

Supposing $x\neq 1$, find

1.  $(2x^2+x-3) \div (x-1)$

2.  $(x^3-1)\div (x-1)$

**Solution**:

2

1.  $(2x^2+x-3) \div (x-1)$ $$\PolyLongDiv[]{s}{t}$$
    $\therefore~(2x^2+x-3) \div (x-1)=2x+3$

2.  $(x^3-1)\div (x-1)$ $$\PolyLongDiv[]{s}{t}$$
    $\therefore~(x^3-1)\div (x-1)=x^2+x+1$

Simplify the following

1.  $\dfrac{ x^4 + x^3 +7 x^2 - 6x +8}{x^2+2x+8}$

2.  $\dfrac{ 8x^3 -10 x^2 - x +8}{x-1}$

**Solution**:

1.    $$\PolyLongDiv[]{s}{t}$$

    $\therefore~\dfrac{ x^4 + x^3 +7 x^2 - 6x +8}{x^2+2x+8}=x^2-x+1.$

2.    $$\PolyLongDiv[]{s}{t}$$

    $\therefore~\dfrac{8x^3 -10 x^2 - x +8}{x-1}=8x^2-2x-3+\dfrac{5}{x-1}.$

<span id="exercise:1:new01" label="exercise:1:new01"></span>

1.  Simplify each of the following

    1.  $(x^3+5x^2+5x-2)\div(x+2)$

    2.  $(5x^3-13x^2+10x-8)\div(-x+2)$

    3.  $(6x^4+13x-19)\div(-x+1)$

2.  For the polynomial $2x^3+7x^2-43x+42$

    1.  Use algebraic division to show that $x-2$ is a factor of it;

    2.  Factorise the polynomial completely.

3.  For the polynomial $12x^3+16x^2-3x-4$

    1.  Use algebraic division to show that $2x+1$ is a factor of it;

    2.  Factorise the polynomial completely.

4.  Simplify each of the following

    1.  $\dfrac{27x^3-9x^2-3x-10}{3x-2}$

    2.  $\dfrac{x^4+3x^3-x^2-x+6}{x+3}$

    3.  $\dfrac{2x^4-5x^3+2x^2+5x-10}{x-2}$

    4.  $\dfrac{7x^4-10x^3+3x^2+3x-3}{x-1}$

    5.  $\dfrac{2x^4+8x^3-5x^2-4x+2}{x^2+4x-2}$
